# Programación estructurada

## <font color = blue>Estructuras de control de flujo condicionales</font>
- ## <font color = limegreen>if</font>, significa "si" (condición).
- ## <font color = limegreen>elif</font>, significa "de lo contrario", es una segunda condición y se evalúa si `if` es `False`
- ## <font color = limegreen>else</font>, significa "si no", ocurre si ninguna condición anterior es `True`
## <font color = red>Los condicionales son las estructuras de control más importantes en toda la programación</font>

# if condicion_1:
> ### Se ejecuta si la condición 1 es verdadera

# elif condicion_2:
> ### Se evalúa solamente si la condición 1 fue falsa

# elif condicion_3:
> ### Se evalúa si las condiciones anteriores fueron falsas

# else:
> ### Se ejecuta cuando ninguna condición anterior fue verdadera
